```python
# because the data is scattered, we arrange it using home feature

home = system[['Season', 'MatchDate', 'HomeTeam', 'AwayTeam', 'FullTimeHomeGoals', 'FullTimeAwayGoals']].copy()

#then we look at it
home.head()
```

```python
# and create an away feature as well

away = system[['Season', 'MatchDate', 'AwayTeam', 'HomeTeam', 'FullTimeAwayGoals', 'FullTimeHomeGoals']].copy()

away.head()
```

```python
# let's rename it 

home = system[[
    'Season',
    'HomeTeam',
    'AwayTeam',
    'FullTimeHomeGoals',
    'FullTimeAwayGoals'
]].copy()

home.head()
```

```python
away = system[[
    'Season',
    'MatchDate',
    'HomeTeam',
    'AwayTeam',
    'FullTimeHomeGoals',
    'FullTimeAwayGoals'
]].copy()

away.head()
```

```python
# we will extend the system 

ext_system = pd.concat([home, away], ignore_index=True)

ext_system.head()
ext_system.shape
```

```python
ext_system.columns
```

```python
ext_system['Result'] = np.where(
    ext_system['Goals For'] > ext_system['Goals Against'], 'Win',
    np.where(
    ext_system['Goals For'] == ext_system['Goals Against'], 'Draw',
    'Lose'
))
```

```python
# we peek through the data

ext_system.head()
```

```python
# we also look for the amount of values in the Result Variable

ext_system['Result'].value_counts()
```

```python
ext_system['Result'].value_counts().plot(kind='bar')
```

Now to properly see the points collected we have to create a 'Points' table 

```python
# creating the points table

ext_system['Points'] = np.where(
    ext_system['Result'] == 'Win', 3,
    np.where(
        ext_system['Result'] == 'Draw', 1,
        0
    ))
```

Now that we have the points, let's treat the team to have it's sequence

```python
# we create a team group sequence
team_groups = ext_system.groupby('Team')

# then let's peek at it with 'Arsenal'
team_groups.get_group('Arsenal').head()
```

```python
# to see the end/last seasons of the system dataset
system[['MatchDate', 'Season']].tail(10)
```

```python
ext_system['Previous 5 matches'] = (
    ext_system.groupby('Team')['Points'].transform(lambda x: x.shift(1).rolling(5).sum())
    )

ext_system.head()
```

```python
team_groups.get_group('Arsenal').head()
```
