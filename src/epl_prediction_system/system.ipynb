```python
# because the data is scattered, we arrange it using home feature
home = system[['Season', 'MatchDate', 'HomeTeam', 'AwayTeam', 'FullTimeHomeGoals', 'FullTimeAwayGoals']].copy()

#then we look at it
home.head()
```

```python
# let's rename it 

home = home.rename(columns={
    "HomeTeam": "Team",
    "AwayTeam": "Opponent Team",
    "FullTimeHomeGoals": "Goals For",
    "FullTimeAwayGoals": "Goals Against"
})
```

```python
# and create an away feature as well

away = system[['Season', 'MatchDate', 'AwayTeam', 'HomeTeam', 'FullTimeAwayGoals', 'FullTimeHomeGoals']].copy()

away.head()
```

```python
away = away.rename(columns={
    "AwayTeam": "Team",
    "HomeTeam": "Opponent Team",
    "FullTimeAwayGoals": "Goals For",
    "FullTimeHomeGoals": "Goals Against"
})
```

```python
# we will extend the system 
ext_system = pd.concat([home, away], ignore_index=True)

ext_system.head()
ext_system.shape
```

```python
ext_system.columns
```

```python
ext_system[ext_system['Team'] == 'Arsenal'].head(10)
```

```python
ext_system['Result'] = np.where(
    ext_system['Goals For'] > ext_system['Goals Against'], 
    'Win',
    np.where(
    ext_system['Goals For'] == ext_system['Goals Against'], 
    'Draw',
    'Lose'
    )
)
```

```python
# creating the points table
ext_system['Points'] = np.where(
    ext_system['Result'] == 'Win', 3,
    np.where(
        ext_system['Result'] == 'Draw', 1,
        0
    )
)
```

```python
ext_system = ext_system.sort_values(['Team', 'MatchDate'])
```

```python
# reset the index

ext_system = ext_system.reset_index(drop=True)
```

```python
ext_system['Previous 5 Points'] = (
    ext_system.groupby('Team')['Points'].transform(lambda x: x.shift(1).rolling(5).sum())
)
```

```python
ext_system[ext_system['Team'] == 'Arsenal'][['MatchDate', 'Team', 'Points', 'Previous 5 Points']].head(10)
```

Now we calculate, how many goals the team scored

```python
ext_system['Previous 5 Goals Scored'] = ext_system.groupby('Team')['Goals For'].transform(lambda x: x.shift(1).rolling(5).sum())
```

```python
ext_system[ext_system['Team'] == 'Arsenal'][['MatchDate', 'Team', 'Goals For', 'Previous 5 Goals Scored']].head(10)
```

We will now calculate for the goals conceded

```python
ext_system['Previous 5 Goals Conceded'] = ext_system.groupby('Team')['Goals Against'].transform(lambda x: x.shift(1).rolling(5).sum())
```

```python
ext_system[ext_system['Team'] == 'Arsenal'][['MatchDate', 'Team', 'Goals Against', 'Previous 5 Goals Conceded']].head(10)
```

```python
ext_system[ext_system['Team'] == 'Arsenal'][['MatchDate', 'Team', 'Result', 'Previous 5 Goals Conceded']].head(10)
```

```python
ext_system[ext_system['Team'] == 'Arsenal'][['MatchDate', 'Team', 'Result', 'Previous 5 Goals Conceded']].head(10)
```


### So as we can already calculate the Goals Scored, Points and Goals Conceded. We then need to calculate the form

```python
ext_system['Win'] = ext_system['Result'].map({
    'Win': 1,
    'Draw': 0,
    'Lose': 0
})
```

```python
# to get the previous 5 wins
ext_system['Previous 5 Wins'] = (ext_system.groupby('Team')['Win'].transform(lambda x: x.shift(1).rolling(5).sum()))
```

```python
ext_system[ext_system['Team'] == 'Arsenal'][['MatchDate', 'Team', 'Result', 'Previous 5 Wins']].head(10)
```

```python
ext_system['Lose'] = ext_system['Result'].map({
    'Win': 0,
    'Draw': 0,
    'Lose': 1
})
```

```python

# to get the previous 5 loses
ext_system['Previous 5 Loses'] = ext_system.groupby('Team')['Lose'].transform(lambda x: x.shift(1).rolling(5).sum())
```

```python
ext_system[ext_system['Team'] == 'Arsenal'][['MatchDate', 'Team', 'Result', 'Previous 5 Loses']].head(10)
```

```python
ext_system['Draw'] = ext_system['Result'].map({
    'Win': 0,
    'Draw': 1,
    'Lose': 0
})
```

```python
# to get the previous 5 draws 
ext_system['Previous 5 Draws'] = ext_system.groupby('Team')['Draw'].transform(lambda x: x.shift(1).rolling(5).sum())
```

```python
ext_system[ext_system['Team'] == 'Arsenal'][['MatchDate', 'Team', 'Result', 'Previous 5 Draws']].head(10)
```

## Training and Testing the Model Using Time-Based Split

```python
# we create our train and test dataframes
train = model_data[model_data['MatchDate'] < '2025-01-01'].copy()

test = model_data[model_data['MatchDate'] >= '2025-01-01'].copy()
```

```python
# we check their shape
print('Train:', train.shape)
print('Test:', test.shape)
```

```python
# we check their dates
print('Train dates:', train['MatchDate'].min(), 'to', train['MatchDate'].max())
print('Test dates:', test['MatchDate'].min(), 'to', test['MatchDate'].max())
```

```python
# let's look at what the 
y_train = train['FullTimeResult']
y_test = test['FullTimeResult']
```

```python
print(y_train.head())
print(y_test.head())
```

```python
# we create the features that can be passed unto x
features = [
    'Previous 5 Points_Home',
    'Previous 5 Wins_Home',
    'Previous 5 Draws_Home',
    'Previous 5 Loses_Home',
    'Previous 5 Goals Scored_Home',
    'Previous 5 Goals Conceded_Home',
    'Previous 5 Points_Away',
    'Previous 5 Wins_Away',
    'Previous 5 Draws_Away',
    'Previous 5 Loses_Away',
    'Previous 5 Goals Scored_Away',
    'Previous 5 Goals Conceded_Away',
]
```

```python
# we pass the features through our train and test dataframe
x_train = train[features]
x_test = test[features]
```

```python
print(x_train.head())
```

```python
x_train.describe()
```

Let's use standardize the outputs for easier understanding

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
scaler.fit(x_train)
```

```python
x_train_scaled = scaler.transform(x_train)
```

```python
x_test_scaled = scaler.transform(x_test)
```

```python
# to see the scaler's mean values
print('Means: ')
scaler.mean_
```

```python
# to see the scaler std values
print('Standard Derivation: ')
scaler.scale_
```

```python
x_train_scaled[:5]
```

Through this scaling, we can finally create our model

```python
#let's use the logisticRegression
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(x_train_scaled, y_train)
```

```python
y_pred = model.predict(x_test_scaled)
```

```python
y_pred[:10]
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

accuracy = accuracy_score(y_pred, y_test)

print('Accuracy: ', accuracy)
```

```python
pd.Series(y_pred).value_counts()
```

```python
y_test.value_counts()
```

```python
confusion_matrix(y_test, y_pred)
```

```python
model.classes_
```

```python
probabilities = model.predict_proba(x_test_scaled)
```

```python
probabilities[:5]
```

```python
print(classification_report(y_test, y_pred))
```

```python
model_data['Draw Difference'] = (
    model_data['Previous 5 Draws_Home'] - model_data['Previous 5 Draws_Away']
)
```

```python
model_data['Goals Scored Difference'] = (
    model_data['Previous 5 Goals Scored_Home'] - model_data['Previous 5 Goals Scored_Away']
)

model_data['Goals Conceded Difference'] = (
    model_data['Previous 5 Goals Conceded_Home'] - model_data['Previous 5 Goals Conceded_Away']
)
```

```python
train = model_data[model_data['MatchDate'] < '2025-01-01'].copy()

test = model_data[model_data['MatchDate'] >= '2025-01-01'].copy()
```

```python
print(train.shape, test.shape)
```

```python
features = [
    'Previous 5 Points_Home',
    'Previous 5 Wins_Home',
    'Previous 5 Draws_Home',
    'Previous 5 Loses_Home',
    'Previous 5 Goals Scored_Home',
    'Previous 5 Goals Conceded_Home',
    'Previous 5 Points_Away',
    'Previous 5 Wins_Away',
    'Previous 5 Draws_Away',
    'Previous 5 Loses_Away',
    'Previous 5 Goals Scored_Away',
    'Previous 5 Goals Conceded_Away',
    'Win Difference',
    'Draw Difference',
    'Lose Difference',
    'Goals Scored Difference',
    'Goals Conceded Difference'
]
```

```python
x_train = train[features]
x_test = test[features]
```

```python
scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
model_v2 = LogisticRegression()

model_v2.fit(x_train_scaled, y_train)
```

```python
new_y_pred = model_v2.predict(x_test_scaled)
```

```python
accuracy = accuracy_score(y_test, new_y_pred)
print(accuracy)
```

```python
print(confusion_matrix(y_test, new_y_pred))
print(classification_report(y_test, new_y_pred))
```
